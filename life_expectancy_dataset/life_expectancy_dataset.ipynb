## Importations and configurations

```python
#library importations

import pandas as pd

import os
import requests
import zipfile
from io import BytesIO

version = 1.0
```

```python
raw = './data/raw'
processed_data = './data/processed'
output_data = './data/output'

output_dataset = './data/processed/life_expectancy_dataset.csv'
```

```python
def download_data(data_source_in):
    """this function get files from url and save it in the raw folder

    Args:
        data_source_in (String): url of the file to download
    """
    response = requests.get(data_source_in)

    if response.status_code == 200:
        # Leer el contenido del archivo ZIP en memoria
        zip_file = BytesIO(response.content)
        
        # Descomprimir el contenido del ZIP directamente en el directorio
        with zipfile.ZipFile(zip_file) as z:
            z.extractall(raw)
```

## Download and read data

```python
#download dataset of women life expectancy
download_data('https://api.worldbank.org/v2/es/indicator/SP.DYN.LE00.FE.IN?downloadformat=csv')
```

```python
#download dataset of men life expectancy
download_data('https://api.worldbank.org/v2/es/indicator/SP.DYN.LE00.MA.IN?downloadformat=csv')
```

## load data

```python
for file in os.listdir(raw):
    print(file)
    if file.startswith('API_SP.DYN.LE00.FE.IN_DS2'):
        df_data_women = pd.read_csv(f'{raw}/{file}', skiprows=4)
        df_data_women = df_data_women.iloc[:, :-1]
    if file.startswith('API_SP.DYN.LE00.MA.IN_DS2'):
        df_data_men = pd.read_csv(f'{raw}/{file}', skiprows=4)
        df_data_men = df_data_men.iloc[:, :-1]
    #load metadata of country codes
    if file.startswith('country_codes_list'):
        df_country_code = pd.read_csv(f'{raw}/{file}',
                                    sep=',',
                                    skiprows=0)
```

## Transform data

```python
#upper all strings

df_data_women = df_data_women.applymap(lambda x: x.upper() if isinstance(x, str) else x)
df_data_men = df_data_men.applymap(lambda x: x.upper() if isinstance(x, str) else x)
df_country_code = df_country_code.applymap(lambda x: x.upper() if isinstance(x, str) else x)
```

## Transform data structure from wide to long table

```python
df_data_women = pd.melt(df_data_women,
                        id_vars=['Country Name', 'Country Code', 'Indicator Name', 'Indicator Code'],
                        var_name='Year',
                        value_vars=df_data_women.iloc[:, :-1].columns,
                        value_name='life_expectancy_women')
```

```python
df_data_men = pd.melt(df_data_men,
                      id_vars=['Country Name', 'Country Code', 'Indicator Name', 'Indicator Code'],
                      var_name='Year',
                      value_vars=df_data_men.iloc[:, :-1].columns,
                      value_name='life_expectancy_men')
```

```python
df_data_women = df_data_women[['Country Name', 
                               'Country Code', 
                               'Year', 
                               'life_expectancy_women']]
```

```python
df_data_men = df_data_men[['Country Code', 
                           'Year', 
                           'life_expectancy_men']]
```

## Union of life expectancy of women and men

```python
df_life_expectancy = pd.merge(df_data_women,
                              df_data_men,
                              on=['Country Code', 'Year'],
                              how='inner')
```

```python
df_life_expectancy.columns = ['country_name', 'country_code', 'year', 'life_expectancy_women', 'life_expectancy_men']

df_life_expectancy = df_life_expectancy[['country_code', 'country_name', 'year', 'life_expectancy_women','life_expectancy_men']]
```

```python
df_life_expectancy = pd.merge(df_life_expectancy,
                              df_country_code,
                              left_on='country_code',
                              right_on='alpha-3',
                              how='inner')
```

```python
df_life_expectancy = df_life_expectancy[['country_code',
                                         'country_name',
                                         'region',
                                         'sub-region',
                                         'intermediate-region',
                                         'year',
                                         'life_expectancy_women',
                                         'life_expectancy_men']]
```

## Set NaN values

```python
df_life_expectancy['life_expectancy_women'].fillna(0, inplace=True)
df_life_expectancy['life_expectancy_men'].fillna(0, inplace=True)
```

```python
df_life_expectancy = df_life_expectancy.sort_values(by=['country_name', 'year'], ascending=True)
```

## Save processed data

```python
df_life_expectancy.to_csv(output_dataset, index=False, float_format="%.2f", sep=';', decimal=',')
```
